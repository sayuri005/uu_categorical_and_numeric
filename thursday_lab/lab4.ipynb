```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from scipy import stats
import random
import seaborn as sns
import math
```

```python
rng = np.random.default_rng(seed=50)
```

```python
#Question 1
k=3
list_num = [0]
for i in range(k+1):
    list_num.append(i/k)
random.choice(list_num)

result=[]
n=5000
for i in range(n):
    each_round = random.choice(list_num)
    result.append(each_round)

df_1 = pd.DataFrame(result)
```

```python
sns.ecdfplot( data=df_1)
```

```python
#2
n=32
data2 = rng.uniform(-np.sqrt(3),np.sqrt(3),n)
samp_mean_sqrt = data2.mean()*np.sqrt(n)
result_2 = []
for i in range(5000):
    each_round_2 =  rng.uniform(-np.sqrt(3),np.sqrt(3),n)
    samp_mean_sqrt_2 = each_round_2.mean()*np.sqrt(n)
    result_2.append(samp_mean_sqrt_2)

df_2 = pd.DataFrame(result_2)
sns.ecdfplot( data=df_2)
```

It approaches a normal distribution. 

```python
#3
p_3=0.5
n=40
data3 = rng.binomial(1, p=p_3, size=n)
df3 = pd.DataFrame(data3)
expected = ((df3[0]==1).sum())-(p_3*n)
divided = expected / np.sqrt(n*p_3*(1-p_3))
result_3 = []
for i in range(5000):
    each_round_3 =  rng.binomial(1, p=p_3, size=n)
    df_3_round = pd.DataFrame(each_round_3)
    each_expected = ((df_3_round[0]==1).sum())-(p_3*n)
    each_divided = each_expected / np.sqrt(n*p_3*(1-p_3))
    result_3.append(each_divided)

df_3 = pd.DataFrame(result_3)
sns.ecdfplot( data=df_3)
```

```python
n=40
p_3=0.1
result_3=[]
for i in range(5000):
    each_round_3 =  rng.binomial(1, p=p_3, size=n)
    df_3_round = pd.DataFrame(each_round_3)
    each_expected = ((df_3_round[0]==1).sum())-(p_3*n)
    each_divided = each_expected / np.sqrt(n*p_3*(1-p_3))
    result_3.append(each_divided)

df_3 = pd.DataFrame(result_3)
sns.ecdfplot( data=df_3)
```

It looks like a normal distribution as n increases. When you just change p, it skews the curve. It skews the curve to the left when you increase the p and the distribution becomes heavy towards the rigth when you decrease the p. 

```python
#5
all =[]
K=20
start=len(range(K))
for x in range(K):
    result_5=[]
    for i in range(5000):
        each_round_5 =  rng.uniform(0,1,K)
        sorted_data_5 = sorted(each_round_5)
        select_num = sorted_data_5[start-1]
        result_5.append(select_num)
    all.append(result_5)
    start = start - 1

for rank in all:
    df_5 = pd.DataFrame(rank)
    sns.ecdfplot(data=df_5)

    
```
