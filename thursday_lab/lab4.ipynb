```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from scipy import stats
import random
import seaborn as sns
import math
```

```python
rng = np.random.default_rng(seed=50)
```

```python
#Question 1
k=3
list_num = [0]
for i in range(k+1):
    list_num.append(i/k)
random.choice(list_num)

result=[]
n=5000
for i in range(5000):
    each_round = random.choice(list_num)
    result.append(each_round)

df_1 = pd.DataFrame(result)
```

```python
sns.ecdfplot( data=df_1)
```
