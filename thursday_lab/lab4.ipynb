```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from scipy import stats
import random
import seaborn as sns
import math
```

```python
rng = np.random.default_rng(seed=50)
```

```python
#Question 1
k=3
list_num = [0]
for i in range(k+1):
    list_num.append(i/k)
random.choice(list_num)

result=[]
n=5000
for i in range(5000):
    each_round = random.choice(list_num)
    result.append(each_round)

df_1 = pd.DataFrame(result)
```

```python
#3
data3 = rng.integers(0,1,size=40,endpoint=True)
df3 = pd.DataFrame(data3)
#df3.count(1)
expected = ((df3[0]==1).sum())-(0.5*1)
divided = expected / np.sqrt(40*0.5*(1-0.5))
result_3 = []
for i in range(5000):
    each_round_3 =  rng.integers(0,1,size=40,endpoint=True)
    df_3_round = pd.DataFrame(each_round_3)
    each_expected = ((df_3_round[0]==1).sum())-(0.5*40)
    each_divided = each_expected / np.sqrt(40*0.5*(1-0.5))
    result_3.append(each_divided)

df_3 = pd.DataFrame(result_3)
sns.ecdfplot( data=df_3)


```

```python
for i in range(5000):
    each_round_3 =  rng.binomial(1,0.8,size=40)
    df_3_round = pd.DataFrame(each_round_3)
    each_expected = ((df_3_round[0]==1).sum())-(0.8*40)
    each_divided = each_expected / np.sqrt(40*0.8*(1-0.8))
    result_3.append(each_divided)

df_3 = pd.DataFrame(result_3)
sns.ecdfplot( data=df_3)
```

Looks like a normal distribution. Even after changing p, the curve still continues to hold its normal shape.
